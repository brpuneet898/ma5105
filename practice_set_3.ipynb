```python
# 1

import math

class NVEC:
    def __init__(self, entries):
        self.entries = list(entries)
        self.n = len(entries)

    def __str__(self):
        return str(self.entries)

    def norm(self):
        n = 0
        for value in self.entries:
            n = n + (value)**2
            n = n**0.5
        return n

    def dot(self, other):
        if self.n != other.n:
            return "lengths do not match"
        else:
            dot_prod = 0
            for i in range(self.n):
                dot_prod = dot_prod + self.entries[i]*other.entries[i]
            return dot_prod

    def __add__(self, other):
        if self.n!=other.n:
            return "lengths do not match"
        else:
            total = []
            for i in range(self.n):
                total.append(self.entries[i] + other.entries[i])
            return NVEC(total)

    def __sub__(self, other):
        if self.n!=other.n:
            return "lengths do not match"
        else:
            total = []
            for i in range(self.n):
                total.append(self.entries[i] - other.entries[i])
            return NVEC(total)

class NMAT:
    def __init__(self, entries):
        self.entries = list(entries)
        self.n = len(entries)

    def __str__(self):
        rows = []
        for row in self.entries:
            rows.append(" ".join(str(x) for x in row))
        return "\n".join(rows)

    def transpose(self):
        new = []
        for j in range(self.n):
            newr = []
            for i in range(self.n):
                newr.append(self.entries[i][j])
            new.append(newr)
        return NMAT(new)

    def multiply(self, other):
        if isinstance(other, NMAT):
            result = []
            for i in range(self.n):
                row = []
                for j in range(self.n):
                    total = 0
                    for k in range(self.n):
                        total = total + (self.entries[i][k]*other.entries[k][j])
                    row.append(total)
                result.append(row)
            return NMAT(result)
        else:
            if self.n != other.n:
                return "lengths do not match"

            result = []
            for i in range(self.n):
                total = 0
                for j in range(self.n):
                    total = total + (self.entries[i][j]*other.entries[j])
                result.append(total)
            return NVEC(result)

v = NVEC([3,4,12])
print(v.norm())

a = NMAT([[1,0,0],[0,1,0],[0,0,1]])
print(a)
```
