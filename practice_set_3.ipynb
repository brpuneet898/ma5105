```python
# 1

import math

class NVEC:
    def __init__(self, entries):
        self.entries = list(entries)
        self.n = len(entries)

    def __str__(self):
        return str(self.entries)

    def norm(self):
        n = 0
        for i in range(self.entries):
            n = n + (self.entries[i])**2
            n = n**0.5
        return n

    def dot(self, other):
        if self.n != other.n:
            return "lengths do not match"
        else:
            dot_prod = 0
            for i in range(self.n):
                dot_prod = dot_prod + self.entries[i]*other.entries[i]
            return dot_prod

    def __add__(self, other):
        if self.n!=other.n:
            return "lengths do not match"
        else:
            total = []
            for i in range(self.n):
                total.append(self.entries[i] + other.entries[i])
            return NVEC(total)

    def __sub__(self, other):
        if self.n!=other.n:
            return "lengths do not match"
        else:
            total = []
            for i in range(self.n):
                total.append(self.entries[i] - other.entries[i])
            return NVEC(total)
```
