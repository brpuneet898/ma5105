```python
import math 

principal = 1000
interest = 6
duration = 1


def calculate_monthly_payment(principal, interest, duration):
    r = interest / 12 / 100
    n = duration * 12

    if r == 0:
        payment = principal / n
    else:
        payment = principal * r * (1 + r) ** n / ((1 + r) ** n - 1)

    return payment

payment = calculate_monthly_payment(500000, 7.5, 20)
print("Monthly payment: ", payment)
```

```python
def f(x):
    return x**3 -x + 2

def root(f, a, b, tol):
    if f(a) * f(b) >= 0:
        print("f(a) and f(b) must have opposite signs.")
        return None

    while (b - a) / 2 > tol:
        c = (a + b) / 2
        if f(c) == 0:
            return c
        elif f(a) * f(c) < 0:
            b = c
        else:
            a = c

    return (a + b) / 2
```
