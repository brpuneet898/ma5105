```python
# 22

def newton_raphson(f, df, x0, tol, max_iter):
    x = x0

    for i in range(1, max_iter + 1):
        fx = f(x)
        dfx = df(x)

        print("Iteration:", i, "x =", x, "f(x) =", fx)

        if dfx == 0:
            print("Error: derivative is zero. Newton-Raphson method terminated.")
            return None

        x_new = x - fx / dfx

        if abs(x_new - x) < tol:
            print("Root found:", x_new)
            return x_new

        x = x_new

    print("Maximum number of iterations exceeded.")
    return None

import math

def v(t):
    return 150 * (1 - math.exp(-0.15 * t)) - 9.81 * t

def dvelocity(t):
    return 22.5 * math.exp(-0.15 * t) - 9.81

t0 = 10

tol = 1e-6
max_iter = 1000

t_max = newton_raphson(v, dvelocity, t0, tol, max_iter)

if t_max is not None:
    if 8 <= t_max <= 12:
        print("The solution lies between 8 and 12 seconds.")
    else:
        print("The solution does not lie between 8 and 12 seconds.")

    print("The rocket reaches its maximum height at t =", t_max, "seconds.")

```
