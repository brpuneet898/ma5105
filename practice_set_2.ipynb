```python
# 1

def cal_bill(n):

    if n < 0:
        return ("Invalid Input")

    total = 0
    if (n>=0) and (n<=100):
        total = 2*n
    elif (n>=101) and (n<=200):
        total = 200 + (n-100)*3.5
    elif (n>=201) and (n<=500):
        total = 550 + (n-200)*5
    elif (n>500):
        total = 2050 + (n-500)*7

    if (total > 2000):
        total = total + (0.1*total)

    return total

a = int(input())
answer = cal_bill(a)
print(answer)
```

```python
# 2

def grade(n):
    grade = ""
    if (n>=90) and (n<=100):
        grade = "A"
    elif (n>=80) and (n<=89):
        grade = "B"
    elif (n>=70) and (n<=79):
        grade = "C"
    elif (n>=60) and (n<=69):
        grade = "D"
    elif (n>=50) and (n<=59):
        grade = "E"
    else:
        grade = "F"

    return grade

def avg(a, b, c):
    return (a+b+c)/3

def pass_or_fail(n, a, b, c):
    pf = False
    if (n>=0) and (n<50):
        pf = False
    else:
        pf = True

    if (a < 40) or (b < 40) or (c < 40):
        pf = False
    else:
        pf = True

    return pf

a = int(input("Maths: "))
b = int(input("Prog: "))
c = int(input("Stats: "))

avg_marks = avg(a,b,c)
print(avg_marks)

v = pass_or_fail(avg_marks, a, b, c)
if v == True:
    print("Pass")
else:
    print("Fail")

g = grade(avg_marks)
print(g) 
```

```python
# 3

def quad(a, b, c):
    d = b**2 - 4*a*c
    r1 = 0
    r2 = 0
    if (a<=0):
        return "can't find roots"

    if d>0:
        print("2 distinct real solutions")
        r1 = (-b + d**0.5) / (2*a)
        r2 = (-b - d**0.5) / (2*a)
    elif d==0:
        print("1 repeated real solutions")
        r1 = (-b + d**0.5) / (2*a)
        r2 = (-b - d**0.5) / (2*a)
    else:
        print("no real roots exist")

    return r1, r2

a = int(input())
b = int(input())
c = int(input())

answer = quad(a, b, c)

print(answer)
```
