```python
# 1

def cal_bill(n):

    if n < 0:
        return ("Invalid Input")

    total = 0
    if (n>=0) and (n<=100):
        total = 2*n
    elif (n>=101) and (n<=200):
        total = 200 + (n-100)*3.5
    elif (n>=201) and (n<=500):
        total = 550 + (n-200)*5
    elif (n>500):
        total = 2050 + (n-500)*7

    if (total > 2000):
        total = total + (0.1*total)

    return total

a = int(input())
answer = cal_bill(a)
print(answer)
```
