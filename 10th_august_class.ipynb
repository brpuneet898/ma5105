```python
def quadratic_roots(a, b, c):
    d = b**2 - 4*a*c
    r1 = (-b + (d**(1/2)))/(2*a)
    r2 = (-b - (d**(1/2)))/(2*a)
    return r1, r2

roots = quadratic_roots(1, -5, 6)
print(roots)
```
