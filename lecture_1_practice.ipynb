```python
# how to estimate e using a while loop

total = 1
count = 1

def fact(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * fact(n - 1)

while (1/fact(count)) > 1e-8:
    total = total + (1/fact(count))
    count = count + 1

print("Estimated value of e:", total)
print("Number of terms used:", count)
```

```python
## finding the roots of a function using the bisection method

def f(x):
    return x**3 - x - 2

a = 1
b = 2

while (b-a) >= 1e-6:
    c = (a+b)/2
    if f(a)*f(c) <= 0:
        b = c
    else:
        a = c

root = (a+b)/2
print("Estimated root of the function:", root)
```

```python
# do a numerical integration using a for loop

def f(x):
    return x**2

n = 100
dx = 1/n

for i in range(n):
    x = i * dx
    total = total + f(x)*dx

print("Estimated value of the integral:", total)
```

```python
# do a numerical integration on a rectangle

def f(x, y):
    return x + y

n = 100
dx = dy = 1/n
total = 0

for i in range(n):
    for j in range(n):
        x = i * dx
        y = j * dy
        total = total + f(x, y) * dx * dy

print("Estimated value of the integral:", total)
```

```python
# write a python program that classifies the roots of the quadratic equation

a = int(input("Enter the coefficient a: "))
b = int(input("Enter the coefficient b: "))
c = int(input("Enter the coefficient c: "))

D = b**2 - 4*a*c

if D > 0:
    print("The equation has two distinct real roots.")
elif D == 0:
    print("The equation has one real root.")
else:
    print("The equation has two complex roots.")

print("The value of the discriminant is:", D)
```
